```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("data/full_2019.csv")

df = df.sample(frac = 0.1)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df['id_mutation'] = df['id_mutation'].apply(lambda code_dept : str(code_dept))
```

```python
from sklearn.impute import KNNImputer
```

```python
df_test = df

object_columns = df_test.select_dtypes(include = ['object']).columns

df_test = pd.get_dummies(df_test , columns = object_columns)

# Initialize a KNNImputer object
imputer = KNNImputer(n_neighbors=5)

# Apply KNN imputation
df_imputed = imputer.fit_transform(df_test)

# Replace missing values with the imputed values
df_test = pd.DataFrame(df_imputed, columns = df.columns)
```
